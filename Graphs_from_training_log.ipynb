```python
import matplotlib.pyplot as plt
import pandas as pd
import matplotlib.ticker as ticker

# Load the training log data
df = pd.read_csv('training_log.csv')

# Create a figure with 3 subplots - each with its own X-axis labels
fig, axes = plt.subplots(3, 1, figsize=(10, 18))

# Subplot 1: Loss Curves
axes[0].plot(df['step'], df['train_loss'], label='Train Loss', color='blue', linewidth=1.5, marker='o', markersize=2)
axes[0].plot(df['step'], df['val_loss'], label='Val Loss', color='orange', linewidth=1.5, linestyle='--', marker='x', markersize=2)
axes[0].set_ylabel('Loss')
axes[0].set_xlabel('Step') # X-axis label on first plot
axes[0].set_title('Training and Validation Loss')
axes[0].legend()
axes[0].grid(True, linestyle=':', alpha=0.6)
axes[0].set_xlim(left=0)
axes[0].set_ylim(bottom=0)
axes[0].xaxis.set_major_locator(ticker.MaxNLocator(nbins=10)) # More frequent step ticks

# Subplot 2: BPC Curves
axes[1].plot(df['step'], df['train_bpc'], label='Train BPC', color='green', linewidth=1.5, marker='o', markersize=2)
axes[1].plot(df['step'], df['val_bpc'], label='Val BPC', color='red', linewidth=1.5, linestyle='--', marker='x', markersize=2)
axes[1].set_ylabel('BPC (Bits Per Character)')
axes[1].set_xlabel('Step') # X-axis label on second plot
axes[1].set_title('Training and Validation BPC')
axes[1].legend()
axes[1].grid(True, linestyle=':', alpha=0.6)
axes[1].set_xlim(left=0)
axes[1].set_ylim(bottom=0)
axes[1].xaxis.set_major_locator(ticker.MaxNLocator(nbins=10))

# Subplot 3: Learning Rate
axes[2].plot(df['step'], df['lr'], label='Learning Rate', color='purple', linewidth=1.5, marker='.', markersize=4)
axes[2].set_ylabel('LR')
axes[2].set_xlabel('Step') # X-axis label on third plot
axes[2].set_title('Learning Rate Schedule')
axes[2].legend()
axes[2].grid(True, linestyle=':', alpha=0.6)
axes[2].set_xlim(left=0)
axes[2].set_ylim(bottom=0)
axes[2].xaxis.set_major_locator(ticker.MaxNLocator(nbins=10))

plt.tight_layout()
plt.savefig('learning_curves_with_steps.png')
plt.show()
```
